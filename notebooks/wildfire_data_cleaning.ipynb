## Spacial Join (Province)

```python
!pip install geopandas
```

```python
import geopandas as gpd
import pandas as pd

# 1. Load batas provinsi (GeoJSON/Shapefile apapun)
provinces = gpd.read_file(r"C:\Users\hanif\Downloads\Data Analyst\Project\end-to-end\Indonesia Deforestation & Wildfire Analysis (2001–2024)\indonesia-deforestation-wildfire-analysis\data\indo_prov34.json")
print(provinces.crs)  # pastikan EPSG:4326 (WGS84) — sama seperti lat/lon biasa

# 2. Load titik hotspot -> ubah jadi GeoDataFrame
df = pd.read_csv(r"C:\Users\hanif\Downloads\Data Analyst\Project\end-to-end\Indonesia Deforestation & Wildfire Analysis (2001–2024)\indonesia-deforestation-wildfire-analysis\data\processed\wildfire_cleaned.csv", usecols=['latitude','longitude'])
uniq = df.drop_duplicates().reset_index(drop=True)

points = gpd.GeoDataFrame(
    uniq,
    geometry=gpd.points_from_xy(uniq['longitude'], uniq['latitude']),  # PENTING: x=lon, y=lat
    crs='EPSG:4326'
)

# 3. Samakan CRS kedua layer (wajib, kalau beda bakal salah hasil)
provinces = provinces.to_crs(points.crs)

# 4. Spatial join
joined = gpd.sjoin(
    points,
    provinces[['NAME_1', 'geometry']],  # ambil kolom nama provinsi + geometry saja
    how='left',
    predicate='within'   # titik "within" (di dalam) polygon
)

joined = joined.rename(columns={'NAME_1': 'province'}).drop(columns=['index_right'])
joined.to_csv(r'C:\Users\hanif\Downloads\Data Analyst\Project\end-to-end\Indonesia Deforestation & Wildfire Analysis (2001–2024)\indonesia-deforestation-wildfire-analysis\data\processed\coord_to_province_lookup.csv', index=False)
```
