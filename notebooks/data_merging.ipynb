## National Dataset Merging

```python
import pandas as pd

country_tc_loss = pd.read_csv(r"C:\Users\hanif\Downloads\Data Analyst\Project\end-to-end\Indonesia Deforestation & Wildfire Analysis (2001–2024)\indonesia-deforestation-wildfire-analysis\data\processed\country_tree_cover_loss.csv")
country_primary_loss = pd.read_csv(r"C:\Users\hanif\Downloads\Data Analyst\Project\end-to-end\Indonesia Deforestation & Wildfire Analysis (2001–2024)\indonesia-deforestation-wildfire-analysis\data\processed\country_primary_loss.csv")
carbon_gross_emissions = pd.read_csv(r"C:\Users\hanif\Downloads\Data Analyst\Project\end-to-end\Indonesia Deforestation & Wildfire Analysis (2001–2024)\indonesia-deforestation-wildfire-analysis\data\processed\carbon_gross_emissions.csv")
```

```python
country_tc_loss.head()
```

```python
country_primary_loss.head()
```

```python
carbon_gross_emissions.head()
```

## Merge Country Tree Cover Loss & Country Primary Loss

```python
national = pd.merge(
    country_tc_loss,
    country_primary_loss,
    on=["country", "year"],
    how="inner"
)

national.head()
```

## Merge National data with carbon_gross_emissions data

```python
national = pd.merge(
    national,
    carbon_gross_emissions,
    on = ["country","year"],
    how = "inner"
)

national.head()
```

```python
national.info()
```

## Save to CSV

```python
national_def = national [[
    "country", "year","tree_cover_loss_ha_x", "tree_cover_loss_ha_y", "gross_emissions_Mg_CO2e"]]
```

```python
national_def.head()
```

```python
national_def = national_def.rename(
    columns={
        "tree_cover_loss_ha_x":"tc_loss_ha",
        "tree_cover_loss_ha_y":"primary_loss_ha"}
)
```
