## Convert Date

```python
import numpy as np

df['acq_date'] = pd.to_datetime(df['acq_date'])
df.info()
```

## Create Year & Month Columns

```python
df['year']=df['acq_date'].dt.year
df['month']=df['acq_date'].dt.month
```

```python
df = df[df['type']==0]
```
